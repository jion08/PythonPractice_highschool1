```python
#p.101
school = '한국고등학교'
grade = 1
name = '김정은'
#문자열 'school'과 변수 school의 값을 출력
print('school', school)
#str을 쓰는 이유는 숫자와 문자열이 섞일수 없기때문에 자료형을 문자열로 바꾼것임.
print(str(grade) + '학년' + name)
```

```python
#p.102
school = '한국'
grade = 1
ban = 3
name = '김정은'
#문자열 포매팅을 사용하지 않은 경우
print(school+'고등학교 '+str(grade)+'학년 '+str(ban)+'반 '+name+'학생, 만나서 반갑습니다.')
#문자열 포매팅을 사용한 경우
print(f'{school}고등학교 {grade}학년 {ban}반 {name}학생, 만나서   ㅜ반갑습니다.')
```
